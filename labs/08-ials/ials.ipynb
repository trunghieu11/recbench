**Exercise.** Do popular items have longer item vectors? Compute the correlation between each item's vector length and its popularity.

<details><summary>Hint</summary>

`np.linalg.norm(model.item_factors, axis=1)` gives the lengths.

</details>

<details><summary>Solution</summary>

```python
import numpy as np

length = np.linalg.norm(model.item_factors, axis=1)
print(round(float(np.corrcoef(length[1:], data.item_pop[1:])[0, 1]), 3))
```
Usually positive: a longer vector gives larger dot products with everyone, a learned popularity.

</details>
